### 2.1 Lists and indexing

A Python list stores an ordered collection of values.

Python uses **zero-based indexing**: the first element has index `0`.

````python
scores = [12, 15, 9, 18, 14]

print("All scores:", scores)
print("First score:", scores[0])
print("Last score:", scores[-1])
print("First three scores:", scores[:3])
````

### 2.2 Dictionaries

A Python dictionary stores data as **key-value pairs**.

Dictionaries are useful when data have meaningful labels rather than numerical positions.

````python
student = {
    "name": "Alice",
    "program": "SBI",
    "grade": 15.5,
    "passed": True
}

print(student)
print("Name:", student["name"])
print("Grade:", student["grade"])
````

````python
student["grade"] = 16
student["year"] = "M1"

print(student)
````

### 2.3 Conditions and loops

Conditions allow a program to make decisions.

Loops allow us to repeat an operation over several values.

````python
scores = [12, 15, 9, 18, 14]

for score in scores:
    if score >= 10:
        print(score, "→ Pass")
    else:
        print(score, "→ Fail")
````

### 2.4 Functions

A function groups reusable instructions.

Functions help us avoid repeating code and make data-processing workflows easier to understand.

````python
def classify_score(score):
    if score >= 10:
        return "Pass"
    else:
        return "Fail"


print(classify_score(15))
print(classify_score(8))
````

### ✏️ Mini-challenge — A first sales report

You are a junior data analyst for an e-commerce company. Your manager gives you the sales amounts (in euros) for six completed orders. Prepare a short report using **plain Python**, before moving to NumPy.

**Work in pairs:** agree on an approach, write your code, then compare and explain your results. Use the refresher examples if needed.

1. Display the number of orders, total sales and average sales per order.
2. Write `classify_sales(value)`: return `"High"` for a value greater than or equal to 150, and `"Standard"` otherwise. Apply it to each amount.
3. Use a loop and a condition to build a list of the amounts classified as `"High"`.
4. Store the number of orders, total and mean in a dictionary called `report`, using the keys `"orders"`, `"total"`, `"mean"`. Print it and explain one result to your partner.

**Self-check:** 6 orders; total €770; average approximately €128.33; high amounts `[180, 160]`. Interpret the average as sales per completed order, not monthly sales or sales per customer.

Do not open the solution until you have tried. The purpose is to diagnose Python fluency, not introduce new syntax.


````python
sales = [120, 95, 140, 180, 75, 160]

# 1. Number of orders, total and mean
number_of_orders = len(sales)
total_sales = sum(sales)
mean_sales = total_sales / number_of_orders
# 2. Define classify_sales(value), then apply it to each amount
def classify_sales(value):
    if value >= 150:
        return "High"
    else:
        return "Standard"
# 3. Build the list of high amounts using a loop and a condition
high_sales = []
for sale in sales:
    if classify_sales(sale) == "High":
        high_sales.append(sale)
# 4. Build and display report
report = {
    "orders": number_of_orders,
    "total": total_sales,
    "mean": mean_sales}
print(report)
print("High sales amounts:", high_sales)
````

## 3. From Python Lists to NumPy Arrays

Python lists are convenient, but data analysis often requires performing numerical operations on many values.

**NumPy** (Numerical Python) provides efficient multidimensional arrays and numerical operations.

Let's compare a Python list with a NumPy array.
**Main new content:** NumPy arrays, indexing, filtering, vectorization and descriptive statistics.

The sales examples describe small fictional business datasets. Their values and observation units are stated separately; they are not all the same dataset.


````python
import numpy as np

sales_list = [120, 135, 150, 165]

sales_array = np.array(sales_list)

print("Python list:", sales_list)
print("NumPy array:", sales_array)
````

````python
# What happens if we multiply by 2?

print("Python list:")
print(sales_list * 2)

print("\nNumPy array:")
print(sales_array * 2)
````

### 3.1 Exploring NumPy arrays

A NumPy array has several useful properties:

- `shape` — dimensions of the array
- `size` — total number of elements
- `dtype` — data type of the elements
- `ndim` — number of dimensions

````python
sales = np.array([120, 135, 128, 150, 165, 172])

print("Array:", sales)
print("Shape:", sales.shape)
print("Size:", sales.size)
print("Data type:", sales.dtype)
print("Number of dimensions:", sales.ndim)
````

### 3.2 Indexing and slicing

NumPy arrays use the same basic indexing principles as Python lists.

We can access individual observations or select subsets of an array.

````python
sales = np.array([120, 135, 128, 150, 165, 172])

print("First value:", sales[0])
print("Last value:", sales[-1])
print("First three values:", sales[:3])
print("Values from index 2 to 4:", sales[2:5])
````

### 3.3 Filtering data

NumPy can select observations that satisfy a condition.

The expression below creates a Boolean array containing `True` or `False` for each observation.

````python
sales >= 150
````

````python
# Use the Boolean mask to select the matching observations
print("Sales >= 150:", sales[sales >= 150])
````

### 3.4 Vectorized operations

NumPy allows us to perform an operation on an entire array without writing an explicit loop.

This is called **vectorization**.

````python
sales = np.array([120, 135, 128, 150, 165, 172])

# Apply a 10% increase
sales_after_increase = sales * 1.10

print("Original sales:", sales)
print("After 10% increase:", sales_after_increase)
````

````python
difference = sales_after_increase - sales

print("Increase:", difference)
````

### 3.5 Descriptive statistics

NumPy provides functions for computing common summary statistics.

These statistics help us obtain a first understanding of a dataset.

````python
sales = np.array([120, 135, 128, 150, 165, 172])

print("Mean:", np.mean(sales))
print("Median:", np.median(sales))
print("Minimum:", np.min(sales))
print("Maximum:", np.max(sales))
print("Standard deviation:", np.std(sales))
````

### ✏️ Your turn — Monthly revenue analysis

A company recorded the following monthly revenues (in thousands of euros):

```python
revenue = np.array([42, 48, 51, 39, 55, 62, 58, 70, 65, 74, 69, 80])
```

Try the six tasks before opening the solution.


1. Display the number of observations.
2. Compute the mean revenue.
3. Find the minimum and maximum revenue.
4. Select all months with revenue greater than or equal to 60.
5. Apply a hypothetical 5% increase to all revenues.
6. Count how many months have revenue greater than or equal to 60.

````python

revenue = np.array([42, 48, 51, 39, 55, 62, 58, 70, 65, 74, 69, 80])

print(revenue.size)
print(np.mean(revenue))
print(np.min(revenue), np.max(revenue))
print(revenue[revenue >= 60])
print(revenue * 1.05)
print(np.sum(revenue >= 60))
````

````python
# Solution

print("Number of observations:", revenue.size)
print("Mean:", np.mean(revenue))
print("Minimum:", np.min(revenue))
print("Maximum:", np.max(revenue))

high_revenue = revenue[revenue >= 60]
print("Revenue >= 60:", high_revenue)

revenue_after_increase = revenue * 1.05
print("After 5% increase:", revenue_after_increase)

print("Number of months >= 60:", np.sum(revenue >= 60))
````
